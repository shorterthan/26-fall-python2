# if-elif-else

```python
def comparison_demo(num1, num2):
    """
    :param num1: number
    :param num2: number
    :return: Returns 1 if num1 is larger, 2 if num2 is larger, or 0 if neither is larger
    """
    if num1 > num2:
        print(num1, 'is greater than', num2)
        return 1
    elif num1 == num2:
        print(num1, 'and', num2, 'are equal')
        return 0
    else:
        print(num1, 'is less than', num2)
        return 2
```

```python
comparison_demo(3, 4)
```

```python
comparison_demo(3, 3)
```

# Practice

Write a function that takes an integer as input.
- Return the integer 3 if the input is divisible by 3 but not 5
- Return the integer 5 if the input is divisible by 5 but not 3
- Return the integer 15 if the input is divisible by both 3 and 5

```python
# Example of checking divisibility with modulus % operator
9 % 3 == 0
```

```python
9 % 5 == 0
```

```python
def three_five_divides(num):
    if (num % 3 == 0) and (num % 5 != 0):
        return 3
    elif (num % 3 != 0) and (num % 5 == 0):
        return 5
    elif (num % 3 == 0) and (num % 5 == 0):
        return 15
    else:
        return 0
```

```python
three_five_divides(15)
```

# for loop

```python
def for_demo(input_int):
    """
    :param input_int: integer
    :return: sum of all integers from 1 to input_int, inclusive
    """
    the_sum = 0
    for current_int in range(1,input_int + 1):
        print('current_int has the value', current_int)
        the_sum = the_sum + current_int
        print('current value of the_sum', the_sum)

    return the_sum
```

```python
result = for_demo(5)
print(result)
```

# Practice

Write a function that takes an integer as input. Use a for loop to return the sum of the squares of all positive integers strictly less than the input.

```python
def square_sum(input_int):
    the_sum = 0
    for current_int in range(1,input_int):
        the_sum = the_sum + current_int**2
    return the_sum
```

```python
square_sum(5)
```

# while loop

```python
def while_demo(input_int):
    """
    :param num1: number
    :return: return the first multiple of 10 that is greater than or equal to input_int
    """
    multiple = input_int
    while multiple % 10 != 0:
        print('multiple is currently:', multiple)
        multiple = multiple + 1

    return multiple
```

```python
result = while_demo(51)
print(result)
```

```python
result = while_demo(10)
print(result)
```

# Practice

Write a function that takes an integer as input. Use a while loop to return the sum of the squares of every even number strictly less than the input.

```python
def square_even(input_int):
    the_sum = 0
    current_int = 2
    while current_int < input_int:
        the_sum = the_sum + current_int**2
        current_int = current_int + 2
        
    return the_sum
```

```python
square_even(5)
```

# More examples

```python
def is_prime(input_int):
    """
    :param input_int: integer
    :return: True if test_int is prime. False, otherwise.
    """
    for test_divisor in range(2, input_int):
        print('currently testing divisibility by:', test_divisor)
        if input_int % test_divisor == 0:
            return False # If you reach this line, you have found a divisor and should stop searching.

    # If you reach this line, then input_int is not divisible by anything in the
    # search range.
    return True
```

```python
result = is_prime(10)
print(result)
```

```python
result = is_prime(35)
print(result)
```

```python
# You can create compound conditional statements using 'and' and 'or'
def either_prime(int1, int2):
    """
    :param int1: integer
    :param int2: integer
    :return: True if at least one of int1 or int2 is prime. False, otherwise.
    """
    if is_prime(int1) or is_prime(int2):
        return True
    
    return False
```

```python
either_prime(3, 5)
```

```python
either_prime(9, 35)
```

```python
def is_perfect(input_int):
    sum = 0
    for i in range(1,input_int):
        if input_int % i == 0:
            sum += i

    if sum == input_int:
        return True
    return False

print(is_perfect(100))
print(is_perfect(496))
```

```python
def lcm(int1, int2):
    if int1 >= int2:
        biggerNum = int1
    else:
        biggerNum = int2


    multiple = biggerNum
    while True:
        if multiple % int1 == 0 and multiple % int2 == 0:
            return multiple
        multiple += biggerNum

print(lcm(6, 10))
print(lcm(3, 10))
print(lcm(10, 30))    

        
```
