```python
def list_demo_nondestructive(input_list):
    """
    :param input_list: a list of integers
    :return: returns a copy of input list with each integer incremented by 1
    """

    # Make a copy of the input list so no changes are made to user's input
    incremented_list = input_list.copy()
    
    # loc is needed as the loop index
    for loc in range(len(incremented_list)):
        incremented_list[loc] = incremented_list[loc] + 1

    return incremented_list
```

```python
def list_demo_destructive(input_list):
    """
    :param input_list: a list of integers
    :return: returns a copy of input list with each integer incremented by 1
    """

    # loc is needed as the loop index
    for loc in range(len(input_list)):
        input_list[loc] = input_list[loc] + 1

    return input_list
```

```python
test_list = [5, 8, 4]
inc_list = list_demo_nondestructive(test_list)
print('Original list:', test_list)
print('Incremented list:', inc_list)
```

```python
def list_demo2(word_list):
    """
    :param word_list: a list of strings (words)
    :return: returns a new list containing all words from input_list that contain an 'e'
    """

    e_list = list()

    # This is the preferred Python syntax for scanning a list when possible
    # Much cleaner style than looping over a range(len(word_list))
    for word in word_list:
        if 'e' in word:
            e_list.append(word)

    return e_list
```

```python
test_list = ['apple', 'banana', 'carrot', 'date']
e_words = list_demo2(test_list)
print(e_words)
```

```python
# Looping over two lists separately
num_list = [1, 2, 3]
string_list = ['a', 'b', 'c']
for num in num_list:
    for letter in string_list:
        print('The num is', num, 'and the letter is', letter)
```

### Practice

Create a function that takes two lists of integers as inputs (call them list1, list2, and list_sum). The function should return True if there is a solution to x + y = z, where x comes from list1, y comes from list2, and z comes from list_sum. Return False, otherwise. For example, list1 = [1, 2], list2 = [0, 1, 3], and list_sum = [5, 6] should return True since you can find 2 + 3 = 5.

```python
def sum_check(list1, list2, list_sum):

    for num1 in list1:
        for num2 in list2:
            sum = num1 + num2
            if sum in list_sum:
                return True

    return False
```

```python
list1 = [1,2]
list2 = [0,1,3]
list_sum = [5,6]
sum_check(list1, list2, list_sum)
```

### Dictionaries 

Use a dictionary when you want to reference the items in the collection (called "values") by something other than their location (called "keys").

```python
def dictionary_demo(upper_int):
    """
    :param upper_int: integer
    :return: a dictionary whose keys are perfect squares and whose values are their
                associated positive square roots
    """

    square_dict = dict()

    for current_int in range(1, upper_int + 1):
        square_dict[current_int**2] = current_int

    return square_dict
```

```python
square_roots = dictionary_demo(10)
print('Dictionary:', square_roots)
print('Keys:', square_roots.keys())
print('Values:', square_roots.values())
print('The square root of 81 is', square_roots[81])
#print('There is no key 0', square_roots[0])
```

```python
# You can check whether keys or values exist in a dictionary
if 81 in square_roots.keys():
    print('81 is a key')
if 9 in square_roots.values():
    print('9 is a value')
if 5 in square_roots:
    print('This wont print')
if 5 in square_roots.values():
    print('5 is a value')
```

### Practice

Write a function that takes two equal-length lists of strings as input (call them key_list and val_list). The function should return a dictionary with the key-value pairs key_list[0] with val_list[0], key_list[1] with val_list[1], and so forth. For example, key_list = ['a', 'b'] and val_list = ['x', 'y'] should return the dictionary {'a':'x', 'b':'y'}.

```python
def create_dict(key_list, val_list):

    new_dict = dict()

    for loc in range(len(key_list)):
        current_key = key_list[loc]
        current_val = val_list[loc]
        new_dict[current_key] = current_val

    return new_dict
```

```python
create_dict(['a','b'], ['x', 'y'])
```

### Sets
Use a set when you don't care about the order of the values. With a set, you only care whether the value exists in the set or not.

```python
def set_demo(num_ints):
    """
    :param num_ints: integer
    :return: print some data about lookup times for lists and sets
    """

    import time

    test_list = list()
    test_set = set()

    for curr_int in range(num_ints):
        test_list.append(curr_int)
        test_set.add(curr_int)

    start = time.time()
    for curr_int in range(num_ints):
        if curr_int in test_list:
            continue
    end = time.time()
    lis_time = end - start
    print('Lookup time for lists:', lis_time)

    start = time.time()
    for curr_int in range(num_ints):
        if curr_int in test_set: # just a little check to waste time
            continue
    end = time.time()
    set_time = end - start
    print('Lookup time for sets:', set_time)
```

```python
# Sets are incredibly fast at lookup compared to other data structures.
set_demo(20000)
```

### Practice

Write a function that takes two sets as input (call them set1 and set2). The function should return True if set1 and set2 have any elements in common. Otherwise, return False. (There are some clever set functions to handle this, but stick to using a for loop to scan over the first set and check whether that element belongs to the other set.)

```python
def has_intersection(set1, set2):

    for element in set1:
        if element in set2:
            return True

    return False
```

```python
has_intersection({1,2,3,4,5}, {5,6,7,8})
```

### Tuples
Tuples have a few specialized use cases.

```python
# 1. They are immutable
tup = (1,2,3)
print(tup[1])

# Will cause an error, since tuples are immutable
#tup[0] = 5
```

```python
# 2. They are traditionally used as heterogeneous data structures while lists are typically homogeneous
tup = ('Austin', 'Mohr', 29) # immutability reliably keeps FirstName, LastName, Age in their correct positions
lis = ['Austin', 'Kristin', 'Brent'] # users expect a list to contain all the same time of data, like FirstNames
```

```python
# 3. They can be used as dictionary keys
age = {('Mohr', 'Austin'): 29, ('of Alexandria', 'Euclid'): 2350}
age[('Cantor', 'Georg')] = 174
print(age)
print(age['Mohr', 'Austin']) # You can omit the parens on a tuple
print(age['of Alexandria', 'Euclid'])
print(age['Cantor','Georg'])
```

```python
# 4. You can loop over two lists at once
list1 = [1,2,3]
list2 = ['a', 'b', 'c']
combined_zip = zip(list1, list2)
print(combined_zip) # an advanced data type that allows for looping

for (number, character) in combined_zip: # loop using tuple syntax
    print('Number', number, 'is associated with character', character)
```

### Practice

Modify your dictionary function to use the zip command: zip the two input lists at the start, then use a tuple to loop over both lists simultaneously and set up the key-value pairs.

```python
def create_dict_zip(key_list, val_list):

    new_dict = dict()

    zipped_pairs = zip(key_list, val_list)
    for (curr_key, curr_val) in zipped_pairs:
        new_dict[curr_key] = curr_val

    return new_dict
```

```python
create_dict_zip(['a','b'], ['x', 'y'])
```
