(Generic) Substitution Cipher
substitution_encrypt(plaintext_file, ciphertext_file, cipher_alphabet): This function should read the contents of plaintext_file, encrypt it using the substitution cipher defined by cipher_alphabet, and write the resulting ciphertext to ciphertext_file. You may decide on the way you wish to represent the ciphertext alphabet, but you should include some error checking to make sure it is valid (i.e., no two letters are encrypted to the same letter). If the alphabet is invalid, terminate the function without producing a ciphertext.
substitution_decrypt(ciphertext_file, plaintext_file, cipher_alphabet): This function should read the contents of ciphertext_file, decrypt it using the substitution cipher defined by cipher_alphabet, and write the resulting plaintext to plaintext_file. You may decide on the way you wish to represent the ciphertext alphabet, but you should include some error checking to make sure it is valid (i.e., no two letters are encrypted to the same letter). If the alphabet is invalid, terminate the function without producing a plaintext.

```python
def substitution_encrypt(plaintext_file, ciphertext_file, cipher_alphabet):
    with open(plaintext_file, 'r') as read_file:
        with open(ciphertext_file, 'w') as write_file:
            for line in read_file:
                for character in line:
                    if character in cipher_alphabet.keys():
                        new_character = cipher_alphabet[character]
                    elif character.upper() in cipher_alphabet.keys():
                        new_character = cipher_alphabet[character.upper()].lower()
                    else:
                        new_character = character
                    write_file.write(new_character)
```
