Substitution encryption, decryption, and cipher alphabet validity checker:

```python
def substitution_encrypt(plaintext_file, ciphertext_file, cipher_alphabet):
    
    if not is_valid_cipher_alphabet(cipher_alphabet):
        return

    with open(plaintext_file, 'r') as read_file:
        with open(ciphertext_file, 'w') as write_file:

            for line in read_file:
                for character in line:

                    if character in cipher_alphabet.keys():
                        new_character = cipher_alphabet[character]

                    elif character.upper() in cipher_alphabet.keys():
                        new_character = cipher_alphabet[character.upper()].lower()

                    else:
                        new_character = character

                    write_file.write(new_character)

def substitution_decrypt(ciphertext_file, plaintext_file, cipher_alphabet):
    if not is_valid_cipher_alphabet(cipher_alphabet):
        return

    cipher_alphabet_keys = cipher_alphabet.keys()
    cipher_alphabet_vals = cipher_alphabet.values()
    cipher_alphabet_reversed = dict(zip(cipher_alphabet_vals, cipher_alphabet_keys))

    with open(ciphertext_file, 'r') as read_file:
        with open(plaintext_file, 'w') as write_file:

            for line in read_file:
                for character in line:

                    if character in cipher_alphabet_vals:
                        new_character = cipher_alphabet_reversed[character].values()

                    elif character.upper() in cipher_alphabet_vals:
                        new_character = cipher_alphabet_reversed[character.upper()].lower()

                    else:
                        new_character = character

                    write_file.write(new_character)

def is_valid_cipher_alphabet(cipher_alphabet):
    
    if len(set(cipher_alphabet)) != len(cipher_alphabet):
        return False

    if len(cipher_alphabet) != 26:
        return False

    return True
```

ROT-13 Cipher (cipher, encryption, and decryption):

```python
def rot13_cipher():

    orig_alph_str = 'ABCDEFGHIJKLMNOPQRSTUVWXYZ'
    cipher_alphabet = dict()

    for i in range(len(orig_alph_str)):
        cipher_alphabet[orig_alph_str[i]] = orig_alph_str[(i + 13)%26]

    return cipher_alphabet

def rot13_encrypt(plaintext_file, ciphertext_file):
    rot13_cipher()
    substitution_encrypt(plaintext_file, ciphertext_file, cipher_alphabet)

def rot13_decrypt(ciphertext_file, plaintext_file):
    rot13_cipher()
    substitution_decrypt(ciphertext_file, plaintext_file, cipher_alphabet)
```

Atbash Cipher (cipher, encryption, and decryption):

```python
def atbash_cipher():

    orig_alph_str = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    cipher_alphabet = dict()

    for i in range(len(orig_alph_str)):
        cipher_alphabet[orig_alph_str[i]] = orig_alph_str[25-i]
        
    return cipher_alphabet

def atbash_encrypt(plaintext_file, ciphertext_file):
    atbash_cipher()
    substitution_encrypt(plaintext_file, ciphertext_file, cipher_alphabet)

def atbash_decrypt(ciphertext_file, plaintext_file):
    atbash_cipher()
    substitution_decrypt(ciphertext_file, plaintext_file, cipher_alphabet)
```
