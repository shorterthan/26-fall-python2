```python
def bubble_sort(arr, asc=True):
    length = len(arr)
    if asc:  # sort ascending
        for iteration in range(length - 1):
            for loc in range(length - 1):
                print(arr)
                if arr[loc] > arr[loc + 1]:
                    (arr[loc], arr[loc + 1]) = (arr[loc + 1], arr[loc])
    else:  # sort descending
        for iteration in range(length - 1):
            for loc in range(length - 1):
                print(arr)
                if arr[loc] < arr[loc + 1]:
                    (arr[loc], arr[loc + 1]) = (arr[loc + 1], arr[loc])

def insertion_sort(arr, asc=True):
    length = len(arr)
    for i in range(1, length):
        current = arr[i]
        loc = i - 1
        if asc:  # sort ascending
            while loc >= 0 and arr[loc] > current:
                arr[loc + 1] = arr[loc]
                loc -= 1
        else:  # sort descending
            while loc >= 0 and arr[loc] < current:
                arr[loc + 1] = arr[loc]
                loc -= 1
        arr[loc + 1] = current

def selection_sort(arr, asc=True):
    length = len(arr)
    for i in range(length - 1):
        selected = i
        for loc in range(i + 1, length):
            if asc:  # sort ascending
                if arr[loc] < arr[selected]:
                    selected = loc
            else:  # sort descending
                if arr[loc] > arr[selected]:
                    selected = loc
        (arr[i], arr[selected]) = (arr[selected], arr[i])

def improved_bubble_sort(arr, asc=True):
    length = len(arr)
    for iteration in range(length - 1):
        swapped = False
        for loc in range(length - 1 - iteration):
            if asc:  # sort ascending
                if arr[loc] > arr[loc + 1]:
                    (arr[loc], arr[loc + 1]) = (arr[loc + 1], arr[loc])
                    swapped = True
            else:  # sort descending
                if arr[loc] < arr[loc + 1]:
                    (arr[loc], arr[loc + 1]) = (arr[loc + 1], arr[loc])
                    swapped = True
        if not swapped:
            break

test_list = [5, 3, 4, 2, 1]
insertion_sort(test_list)
print(test_list)

test_list = [5, 3, 4, 2, 1]
selection_sort(test_list, False)
print(test_list)

test_list = [5, 3, 4, 2, 1]
improved_bubble_sort(test_list)
print(test_list)
```
